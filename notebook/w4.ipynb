```python
# ADNI pipeline: CSF tau by diagnosis
Question: does CSF tau differ between CN, MCI and AD?
```

```python
from csv_labeler.config import RAW_DIR, PROCESSED_DIR
from csv_labeler.functions import load_all
from csv_labeler.filter import keep_columns, drop_shared_ids
from csv_labeler.compute import group_summary
from csv_labeler.stats import compare_groups
from csv_labeler.plot import plot_groups
```

```python
df = load_all(RAW_DIR)
df.shape
```

```python
df = keep_columns(df, ["PTID", "label", "MMSCORE", "AB42_FINAL", "TAU_FINAL", "PTAU_FINAL"])
df = drop_shared_ids(df)
df["label"].value_counts().sort_index()
```

```python
group_summary(df, "TAU_FINAL")
```

```python
h, p = compare_groups(df, "TAU_FINAL")
print(f"Kruskal-Wallis: H = {h:.1f}, p = {p:.2e}")
```

```python
fig, ax = plot_groups(df, "TAU_FINAL", ylabel="CSF tau (pg/mL)", title="CSF tau by diagnosis")
fig.savefig(PROCESSED_DIR / "tau_by_group.png", dpi=150, bbox_inches="tight")
```
