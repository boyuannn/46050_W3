# Test load_and_label
Load each CSV, add a label column (CN=0, MCI=1, AD=2), and check the result.

import pandas as pd
from load_data import load_and_label

```python
cn  = load_and_label("data/CN.csv", 0)
mci = load_and_label("data/MCI.csv", 1)
ad  = load_and_label("data/AD.csv", 2)
ad.shape
```

```python
# test 1: The number of lines remains unchanged (the function should not delete or add any lines)
raw_ad = pd.read_csv("data/AD.csv")
assert len(ad) == len(raw_ad)

# test 2: The useless row numbers and columns have been deleted
assert "Unnamed: 0" not in ad.columns

# test 3: The labels on each line are correct
assert (cn["label"] == 0).all()
assert (mci["label"] == 1).all()
assert (ad["label"] == 2).all()

print("All tests passed")
```
